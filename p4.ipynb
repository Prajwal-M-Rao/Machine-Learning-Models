```python
import numpy as np

# Sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Derivative of the sigmoid function
def sigmoid_derivative(x):
    return x * (1 - x)

# Mean squared error loss function
def mse(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

# Neural Network class
class NeuralNetwork:
    def __init__(self, ins, his, ous):
        self.ins = ins
        self.his = his
        self.ous = ous

        self.weights_in_hid = np.random.randn(self.ins, self.his)
        self.bias_hi = np.random.randn(self.his)

        self.weights_hi_ou = np.random.randn(self.his, self.ous)
        self.bias_ou = np.random.randn(self.ous)

    def feed_forward(self, X):
        self.hidden_input = np.dot(X, self.weights_in_hid) + self.bias_hi
        self.hidden_output = sigmoid(self.hidden_input)

        self.final_input = np.dot(self.hidden_output, self.weights_hi_ou) + self.bias_ou
        self.final_output = sigmoid(self.final_input)

        return self.final_output

    def back_propagation(self, X, Y, output, learning_rate):
        output_error = Y - output
        output_delta = output_error * sigmoid_derivative(output)

        hidden_error = output_delta.dot(self.weights_hi_ou.T)
        hidden_delta = hidden_error * sigmoid_derivative(self.hidden_output)

        self.weights_hi_ou += self.hidden_output.T.dot(output_delta) * learning_rate
        self.bias_ou += np.sum(output_delta, axis=0) * learning_rate

        self.weights_in_hid += X.T.dot(hidden_delta) * learning_rate
        self.bias_hi += np.sum(hidden_delta, axis=0) * learning_rate

    def train(self, X, Y, epochs, learning_rate):
        for epoch in range(epochs):
            output = self.feed_forward(X)
            self.back_propagation(X, Y, output, learning_rate)
            if epoch % 1000 == 0:
                loss = mse(Y, output)
                print(f"Epoch {epoch}, Loss: {loss}")

if __name__ == "__main__":
    # Input data
    X = np.array([[2, 9], [1, 5], [3, 6]], dtype=float)
    Y = np.array([[92], [86], [89]], dtype=float)

    # Normalize the data
    X = X / np.max(X, axis=0)
    Y = Y / 100  # Normalizing the target output for the example

    # Create neural network
    nn = NeuralNetwork(ins=2, his=2, ous=1)

    # Train the neural network
    nn.train(X, Y, epochs=10000, learning_rate=0.1)

    # Test the neural network
    output = nn.feed_forward(X)
    print("Predicted Output:")
    print(output * 100)  # De-normalize the output for comparison
```
