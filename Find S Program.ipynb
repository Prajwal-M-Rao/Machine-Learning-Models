```python
import pandas as pd
hypo=['%']*6
```

```python
data = pd.read_csv("trainingdata.csv")
print("\n The given training examples are  \n")
print(data)
```

```python
#filtiring the positive exaples

positive_examples=data[data.iloc[:,-1]=='Yes'] # : represents all rows
positive_examples
```

```python
#initializing the 1st positive as an hypotesis

hypo=positive_examples.iloc[0,:-1].tolist() #tolist() is used group the attributes in a list
print('Initial hypothesis \n',hypo)
```

```python
#Implementing the FIND-S algorithm

for i in range (1,len(positive_examples)):
    print("\nIteration ",i)
    print("Current hpothesis :",hypo)
    for j in range(len(hypo)):
        if hypo[j]!=positive_examples.iloc[i,j]:
            hypo[j]='?'
    print("Updated hypothesis :",hypo)
```
